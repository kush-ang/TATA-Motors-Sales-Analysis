```python
import pandas as pd
```

```python
df=pd.read_csv("TATA_car_sales.csv")
```

```python
df.head(10)
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["promotion"] = df["promotion"].fillna("No Promotion")
```

```python
df.loc[df['is_first_time_buyer'] == True, ['previous_car_brand', 'previous_car_model']] = 'Not Applicable'
```

```python
df["previous_car_brand"] = df["previous_car_brand"].fillna("Unknown")
df["previous_car_model"] = df["previous_car_model"].fillna("Unknown")
```

```python
df.isnull().sum()
```

```python
object_columns = df.select_dtypes(include=['object']).columns
for col in object_columns:
    df[col] = df[col].astype(str).str.strip().str.title()
```

```python
print(df[df["average_sale_price"]<=0])
print(df[df["dealer_discount_pct"]<=0])
```

```python
df.to_csv("TATA_car_sales_cleaned.csv",index=False)
print("Congratulations! Data Cleaning is completed successfully.")
```
